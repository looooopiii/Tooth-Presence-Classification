```python
import os
import pandas as pd

# load csv
df = pd.read_csv("label_flipped.csv")

TEST_PLY_DIR = "/home/user/tbrighton/blender_outputs/parsed_ply"

# normalize df ids
df["new_id"] = df["new_id"].astype(str)

# list ply base names
ply_names = {
    os.path.splitext(f)[0]
    for f in os.listdir(TEST_PLY_DIR)
    if f.endswith(".ply")
}

# find matching names
matching = set(df["new_id"]).intersection(ply_names)

print(f"Total df rows: {len(df)}")
print(f"Matching names: {len(matching)}")
```

```python
df_matched = (
    df[df["new_id"].isin(matching)]
    .drop_duplicates(subset="new_id", keep="first", ignore_index=True)
)

print("Final df rows:", len(df_matched))  # ✅ 167
```

```python
import numpy as np

# make sure string + lowercase
df_matched["new_id"] = df_matched["new_id"].astype(str).str.lower()

df_matched["jaw"] = np.where(
    df_matched["new_id"].str.contains("upper"),
    0,
    np.where(
        df_matched["new_id"].str.contains("lower"),
        1,
        np.nan
    ).astype(int)
)
```

```python
df_matched.to_csv('label_flipped_filtered.csv', index=False)
```
